```python
import pandas as pd, json, sqlite3
df = pd.read_csv("tmdb_5000_movies.csv")
df.head()
df.info()
df.isna().sum() #conta quantos valores (NaN) existem em cada coluna
```

```python
df['release_date'] = pd.to_datetime(df['release_date'])
```

```python
df["ano"] = df["release_date"].dt.year
```

```python
con = sqlite3.connect("filmes.db")
# abre (ou cria) o arquivo do banco

df.to_sql("filmes", con, if_exists="replace", index=False)
# grava o seu df no banco, numa tabela chamada "filmes"

pd.read_sql("""
SELECT title, ano, vote_average
FROM filmes
WHERE genres LIKE '%Science Fiction%'
ORDER BY vote_average DESC
LIMIT 10
""", con)
# pergunta ao banco: "me dê título, ano e nota dos filmes
# cujo gênero contém Science Fiction, do melhor para o pior, só 10"
```

```python
pd.read_sql("SELECT * FROM filmes LIMIT 5", con)
```
